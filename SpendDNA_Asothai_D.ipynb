```python
def extract_vendor(description):
    text = str(description).upper()

    vendors = {
        'SWIGGY': 'Swiggy',
        'ZOMATO': 'Zomato',
        'ZEPTO': 'Zepto',
        'AMAZON': 'Amazon',
        'FLIPKART': 'Flipkart',
        'UBER': 'Uber',
        'OLA': 'Ola',
        'RAPIDO': 'Rapido',
        'NETFLIX': 'Netflix',
        'SPOTIFY': 'Spotify',
        'ZERODHA': 'Zerodha',
        'PHONEPE': 'PhonePe',
        'GPAY': 'Google Pay',
        'GOOGLE PAY': 'Google Pay',
        'PAYTM': 'Paytm'
    }

    for key, vendor in vendors.items():
        if key in text:
            return vendor

    return 'Uncategorised'


df['Vendor'] = df['Description'].apply(extract_vendor)

print(df[['Description', 'Vendor']].head(20))
```

```python
vendor_counts = df['Vendor'].value_counts()

print(vendor_counts)
```

```python
print("Top 10 Vendors")
print(df['Vendor'].value_counts().head(10))
```

```python
def assign_category(row):
    text = str(row['Description']).upper()
    vendor = str(row['Vendor']).upper()

    if vendor in ['SWIGGY', 'ZOMATO']:
        return 'Food & Dining'
    elif vendor == 'ZEPTO':
        return 'Groceries'
    elif vendor in ['AMAZON', 'FLIPKART']:
        return 'Shopping'
    elif vendor in ['UBER', 'OLA', 'RAPIDO']:
        return 'Transport'
    elif vendor in ['NETFLIX', 'SPOTIFY']:
        return 'Entertainment'
    elif vendor == 'ZERODHA':
        return 'Investments'
    elif any(word in text for word in ['RENT', 'HOUSE RENT']):
        return 'Rent'
    elif any(word in text for word in ['ELECTRICITY', 'EB', 'WATER', 'BILL']):
        return 'Bills & Utilities'
    elif any(word in text for word in ['HOSPITAL', 'PHARMACY', 'MEDICAL']):
        return 'Healthcare'
    elif any(word in text for word in ['TRANSFER', 'NEFT', 'IMPS']):
        return 'Personal Transfer'
    else:
        return 'Other'


df['Category'] = df.apply(assign_category, axis=1)

print(df[['Description', 'Vendor', 'Category']].head(20))
```

```python
category_spending = (
    df[df['Type'] == 'Debit']
    .groupby('Category')['Amount']
    .sum()
    .sort_values(ascending=False)
)

print("Category-wise Spending")
print(category_spending)
```

```python
df['Month'] = df['Date'].dt.strftime('%b')

monthly_spending = (
    df[df['Type'].str.lower() == 'debit']
    .groupby('Month')['Amount']
    .sum()
)

print("===== MONTHLY SPENDING =====")
print(monthly_spending)
```

```python
df['Hour'] = pd.to_numeric(
    df['Time'].astype(str).str[:2],
    errors='coerce'
)

def time_period(hour):
    if pd.isna(hour):
        return 'Unknown'
    elif hour < 6:
        return 'Late Night'
    elif hour < 12:
        return 'Morning'
    elif hour < 17:
        return 'Afternoon'
    elif hour < 21:
        return 'Evening'
    else:
        return 'Night'

df['Time_Period'] = df['Hour'].apply(time_period)

time_spending = (
    df[df['Type'].str.lower() == 'debit']
    .groupby('Time_Period')['Amount']
    .sum()
    .sort_values(ascending=False)
)

print("===== TIME-OF-DAY SPENDING =====")
print(time_spending)
```

```python
df['Amount'] = (
    df['Amount']
    .astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace('Rs.', '', regex=False)
    .str.replace('Rs', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
)

df['Amount'] = pd.to_numeric(df['Amount'], errors='coerce')

print("Amount converted successfully!")
print(df['Amount'].dtype)
print(df['Amount'].head())
```

```python
# AI-assisted: basic debugging and z-score calculation syntax.

debit_df = df[df['Type'].astype(str).str.strip().str.lower() == 'debit'].copy()

category_mean = debit_df.groupby('Category')['Amount'].transform('mean')
category_std = debit_df.groupby('Category')['Amount'].transform('std')

category_std = category_std.replace(0, pd.NA)

debit_df['Z_Score'] = (
    (debit_df['Amount'] - category_mean) / category_std
)

anomalies = debit_df[debit_df['Z_Score'] > 2].copy()

print("===== TOP ANOMALIES =====")
print(
    anomalies[
        ['Date', 'Description', 'Category', 'Amount', 'Z_Score']
    ]
    .sort_values('Z_Score', ascending=False)
    .head(10)
)
```

```python
# AI-assisted: rule-based archetype classification syntax.

debit_df = df[
    df['Type'].astype(str).str.strip().str.lower() == 'debit'
].copy()

category_spending = (
    debit_df.groupby('Category')['Amount']
    .sum()
    .sort_values(ascending=False)
)

total_spending = debit_df['Amount'].sum()

top_category = category_spending.index[0]
top_percentage = (category_spending.iloc[0] / total_spending) * 100

late_night_spending = debit_df[
    debit_df['Time_Period'].isin(['Late Night', 'Night'])
]['Amount'].sum()

late_night_percentage = (
    late_night_spending / total_spending
) * 100

print("===== SPENDING ARCHETYPE =====")
print("Top Category:", top_category)
print(f"Top Category Share: {top_percentage:.2f}%")
print(f"Late Night/Night Spending: {late_night_percentage:.2f}%")

if top_category == 'Food & Dining':
    archetype = 'Foodie'
elif top_category == 'Shopping':
    archetype = 'Shopaholic'
elif top_category == 'Transport':
    archetype = 'Cab Commuter'
elif top_category == 'Entertainment':
    archetype = 'Subscription Lover'
elif top_category == 'Investments':
    archetype = 'Investor'
else:
    archetype = 'Balanced Spender'

print("Spending Archetype:", archetype)
```

```python
# AI-assisted: final report formatting and summary syntax.

print("=" * 50)
print("              SpendDNA FINAL REPORT")
print("=" * 50)

print(f"Total Transactions : {len(df)}")
print(f"Total Spending    : ₹{debit_df['Amount'].sum():,.2f}")
print(f"Total Income      : ₹{df[df['Type'].astype(str).str.lower() == 'credit']['Amount'].sum():,.2f}")

print("\nTop 5 Spending Categories:")
print(category_spending.head(5))

print("\nTop Vendors:")
print(df['Vendor'].value_counts().head(5))

print("\nTop Anomalies:")
print(
    anomalies[['Description', 'Category', 'Amount', 'Z_Score']]
    .sort_values('Z_Score', ascending=False)
    .head(5)
)

print("\nSpending Archetype:", archetype)

print("\n===== 3 DATA-SPECIFIC INSIGHTS =====")

print(f"1. The highest spending category is {top_category}, contributing {top_percentage:.2f}% of total debit spending.")

print(f"2. Late Night/Night transactions contribute {late_night_percentage:.2f}% of total spending.")

print(f"3. The detected spending archetype is {archetype} based on the user's spending pattern.")

print("=" * 50)
```
